El parámetro normalize nos ayuda a obtener probabilidades en lugar de frecuencias, además de poder realizar el condicionamiento de una variable sobre otra.

Con el parámetro margins también podemos calcular las probabilidades marginales.

```python
pd.crosstab(datos_completos.layer, datos_completos.aptitud_forVII_lingue, normalize='all', margins=True) # probabilidades, con esto ya podemos saber si los sucesos son independientes 
```

```python
pd.crosstab(datos_completos.layer, datos_completos.aptitud_agricola_lingue, normalize = "all", margins = True)
```

```python
import scipy.stats
import pandas as pd

chi2, p, dof, ex = scipy.stats.chi2_contingency(pd.crosstab(datos_completos.layer, datos_completos.aptitud_forVII_lingue))
print(chi2) # resultado
print(p)    # p-value
print(dof)  # grados de libertad 
print(ex)   # ex las frecuencias esperadas a partir de la tabla de contingencia

## Trabajar en este test para incorporarlo en al streaming
```
